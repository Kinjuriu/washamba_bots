# Herd-Safe v3: Four-Turn Forecast — Notebook Version 2

This updates the existing Herd-Safe notebook with the exact frozen forecast4 candidate. The URL/title retain the historical Herd-Safe v3 name; this is notebook version 2, not a new notebook.

Based on [Dmitrii Gluzdov's Herd-Safe Sale Window, version 2](https://www.kaggle.com/code/dmitriigluzdov/kaggriculture-herd-safe-sale-window-lb-2700). Original Apache-2.0 LICENSE and NOTICE are preserved. This derivative is not an official upstream release.

The active opponent-sale predictor now looks up to four turns ahead instead of two, but additional distant signals require at least three matching historical events for that product and at least 70% matched-event precision over the preceding 240 turns, using the existing +/-1-turn matching tolerance. Immediate one/two-turn signals retain their behavior. This filter is not a calibrated probability. Own sale lookahead (48) and race reservation (44–48) are unchanged. The original v3 risk-feed experiment remains included.

**Decision: locally supported candidate on a small independent panel; leaderboard improvement is not established.**


## Bounded local evidence

Engine: kaggle-environments 1.32.7; full 720-step games, paired seeds, both seats. Screening: 72 games across six policies, three opponents, two fresh seeds. Forecast4 and forecast6 tied on match outcomes; forecast4 was selected by the predeclared relative-cash-margin tiebreak. The source was frozen before independent confirmation. No tuning followed confirmation.

Confirmation: 72 total games = 24 forecast4 games + 24 published-v3 control games + 24 original-v2 control games, on four new seeds 963200–963203. These are not 72 independent candidate trials.

| Opponent | Forecast4 W/L/D | Published v3 W/L/D | Original v2 W/L/D |
|---|---|---|---|
| Herd-Safe v2 | 6/2/0 | 0/4/4 | 0/0/8 |
| More Wheat v9 | 6/2/0 | 0/8/0 | 0/8/0 |
| Order Book v3 | 8/0/0 | 8/0/0 | 8/0/0 |
| Total | 20/4/0 | 8/12/4 | 8/8/8 |

Paired versus published v3: 12 improved, 0 worsened, 12 unchanged outcomes; +41.67 percentage points match points. Versus original v2: 12 improved, 2 worsened, 10 unchanged; +33.33 points. Average paired relative-cash-margin gains: +1167.08 vs v3 and +1166.58 vs v2. Positive average match-point effect on three of four new seeds versus each control. The predeclared practical gate passed.

All 144 screening+confirmation games completed without recorded game errors. Frozen hashes and disjoint seeds verified. Seven saved replays reproduced all 5033 decisions; inspected counters had no internal errors. Four focused semantic tests passed.

## Limitations and rejected hypotheses

Only four independent confirmation worlds and three related public opponents were used. Two pairs worsened versus v2; universal safety and population superiority are not proved. Outcome dependence, opponent ancestry and changing public matchmaking limit leaderboard extrapolation. The prior v3 feeding change alone had not established superiority over v2.

Scenario-robust ordering lost all 12 screening matches and was rejected. Physically confirmed feeding produced no final-cash improvement; audited behavior mostly abstained. Forecast6 did not beat forecast4 in screening. None of those alternatives is combined into this build. No larger sample was run to chase significance.

```python
# Apache-2.0. Confidence gate applies only to additional distant signals.
_HP_ENTRY = [v for v in list(globals().values()) if callable(v)][-1]
_HP_STATS = dict(extension_signals=0,rejected_signals=0)
def _hp_quantity(ev,item,step,seen):
    immediate=sum(ev.get((step+d,item),0) for d in (1,2))
    if immediate>=_V92_P_K: return immediate
    extended=sum(ev.get((step+d,item),0) for d in range(1,_HP_WINDOW+1))
    if extended<_V92_P_K: return immediate
    prior=[(t,i) for (t,i),q in ev.items() if step-240<=t<step-1 and i==item and q>=2]
    hits=sum(any((t+d,i) in seen for d in (-1,0,1)) for t,i in prior)
    if hits>=3 and hits>=0.7*len(prior):
        _HP_STATS['extension_signals']+=1
        return extended
    _HP_STATS['rejected_signals']+=1
    return immediate

_HP_WINDOW = 4
def _v92_predict(obs, action, st):
    step = int(obs["step"])
    _v92_p_update(obs, st)
    if step < 150 or step >= 700:
        return action
    if step % _V92_P_EVERY == 0 or "best" not in st:
        st["best"] = _v92_p_forecast(obs, st)
    best = st["best"]
    if not best:
        return action
    native = _IMPL.chassis.players.get(int(obs["player"]))
    if not native or native.get("route") not in _IMPL.chassis.routes:
        return action
    tape = _IMPL.chassis.routes[native["route"]]
    market = [list(o) for o in action.get("market") or []]
    already = {o[1] for o in market if len(o) > 1 and o[0] in ("SELL", "BUY_PRODUCT")}
    stock = projected_shed(action, FarmView(obs))
    changed = False
    for i, item in enumerate(_V92_P_ITEMS):
        if item not in _V92_P_USE or item in already or len(market) >= MAX_ORDERS:
            continue
        votes = sum(1 for ev in best if _hp_quantity(ev, i, step, st["obs"]) >= _V92_P_K)
        if votes < 1:
            continue
        ours = 0
        for t in range(step + 1, min(len(tape), step + _V92_P_H + 1)):
            ours += sum(min(100, int(o[2])) for o in (tape[t] or {}).get("market") or []
                        if len(o) >= 3 and o[0] == "SELL" and o[1] == item)
        qty = min(int(stock.get(item, 0)), ours)
        if qty > 0:
            market.insert(0, ["SELL", item, qty])
            _V92_P_REPORT["pred_units"] += qty
            _V92_P_REPORT["pred_fires"] += 1
            changed = True
    if not changed:
        return action
    result = dict(action)
    result["market"] = market[:MAX_ORDERS]
    return result

def herdsafe_forecast_agent(observation,configuration=None):
    if int(observation['step'])==0:
        for k in _HP_STATS: _HP_STATS[k]=0
    action=_HP_ENTRY(observation,configuration)
    if int(observation['step'])==718: print('HP_TELEMETRY',_HP_STATS)
    return action
herdsafe_forecast_agent.telemetry=_HP_STATS

```
