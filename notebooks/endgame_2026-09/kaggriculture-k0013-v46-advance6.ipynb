## Pack `submission.tar.gz`

Single-root `main.py`. Stored gzip so the bytes do not depend on zlib.


```python
import binascii
import hashlib
import io
import json
import struct
import tarfile
from pathlib import Path

EXPECTED_MAIN = "45628c719dc967f81655c19f70e579c55a5758b9fe75a5fcdad9193eebf6a017"


def archive_bytes(files: dict[str, bytes]) -> bytes:
    buffer = io.BytesIO()
    with tarfile.open(fileobj=buffer, mode="w", format=tarfile.USTAR_FORMAT) as archive:
        for name in sorted(files):
            entry = tarfile.TarInfo(name)
            entry.size = len(files[name])
            entry.mtime = entry.uid = entry.gid = 0
            entry.uname = entry.gname = ""
            entry.mode = 0o644
            archive.addfile(entry, io.BytesIO(files[name]))
    raw = buffer.getvalue()
    result = bytearray(b"\x1f\x8b\x08\x00\x00\x00\x00\x00\x00\xff")
    for offset in range(0, len(raw), 65535):
        chunk = raw[offset:offset + 65535]
        result.append(1 if offset + len(chunk) == len(raw) else 0)
        result.extend(struct.pack("<HH", len(chunk), len(chunk) ^ 0xffff))
        result.extend(chunk)
    result.extend(struct.pack("<II", binascii.crc32(raw) & 0xffffffff, len(raw) & 0xffffffff))
    return bytes(result)

main_bytes = Path("main.py").read_bytes()
main_sha = hashlib.sha256(main_bytes).hexdigest()
assert main_sha == EXPECTED_MAIN, (main_sha, EXPECTED_MAIN)
payload = archive_bytes({"main.py": main_bytes})
Path("submission.tar.gz").write_bytes(payload)
receipt = {
    "main_sha256": main_sha,
    "archive_sha256": hashlib.sha256(payload).hexdigest(),
    "main_bytes": len(main_bytes),
    "selected_ref": "56306096",
    "selected_public": 2644.2,
}
Path("receipt.json").write_text(json.dumps(receipt, indent=2) + "\n")
print(json.dumps(receipt, indent=2))
```

Confirms `agent` imports. Not a leaderboard forecast.


```python
import importlib.util
from pathlib import Path

spec = importlib.util.spec_from_file_location("k0013_agent", Path("main.py"))
mod = importlib.util.module_from_spec(spec)
spec.loader.exec_module(mod)
assert hasattr(mod, "agent"), "main.py must export agent(obs)"
print("agent loaded:", mod.agent)
```
